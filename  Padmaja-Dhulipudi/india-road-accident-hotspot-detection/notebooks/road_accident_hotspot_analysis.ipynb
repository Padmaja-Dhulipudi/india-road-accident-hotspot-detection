```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Indian Road Accident Data.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", data.shape)
```

```python

print("COLUMN NAMES:")
print(data.columns.tolist())

print("\nFIRST FIVE RECORDS:")
display(data.head())

print("\nDATA TYPES:")
print(data.dtypes)

print("\nMISSING VALUES:")
print(data.isnull().sum())

print("\nDUPLICATES:")
print(data.duplicated().sum())

print("\nSTATISTICAL SUMMARY:")
display(data.describe(include="all"))
```

```python

# STAGE 2: DATA QUALITY CHECK

# 1. Check city-wise accident distribution
print("ACCIDENTS BY CITY")
display(data["city"].value_counts())

# 2. Check city and state combinations
print("\nCITY-STATE COMBINATIONS")
display(
    data.groupby(["city", "state"]).size()
    .reset_index(name="accident_count")
)

# 3. Check geographical coordinate ranges
print("\nCOORDINATE RANGES")
print("Latitude:", data["latitude"].min(), "to", data["latitude"].max())
print("Longitude:", data["longitude"].min(), "to", data["longitude"].max())

# 4. Check unique geographical locations
print("\nUnique coordinate pairs:",
      data[["latitude", "longitude"]].drop_duplicates().shape[0])

# 5. Check date validity
data["date_parsed"] = pd.to_datetime(
    data["date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("\nInvalid dates:", data["date_parsed"].isna().sum())

# 6. Check festival data
print("\nFestival values:")
print(data["festival"].value_counts(dropna=False))

# 7. Plot accidents by city
data["city"].value_counts().plot(
    kind="bar",
    figsize=(10, 5),
    title="Accident Records by City"
)

plt.xlabel("City")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
accidents = data[
    [
        "accident_id",
        "city",
        "state",
        "latitude",
        "longitude",
        "date",
        "hour",
        "accident_severity"
    ]
].copy()
accidents = accidents.dropna(
    subset=["latitude", "longitude"]
)
accidents = accidents[
    accidents["latitude"].between(-90, 90) &
    accidents["longitude"].between(-180, 180)
]
accidents["date"] = pd.to_datetime(
    accidents["date"],
    format="%d-%m-%Y",
    errors="coerce"
)
accidents = accidents.dropna(subset=["date"])
accidents = accidents.reset_index(drop=True)
print("Data preparation completed!")
print("Prepared dataset shape:", accidents.shape)
display(accidents.head())
print("\nMissing values:")
print(accidents.isnull().sum())
```

```python
from sklearn.cluster import DBSCAN
import numpy as np
coordinates = accidents[
    ["latitude", "longitude"]
].to_numpy()
coordinates_rad = np.radians(coordinates)
print("Coordinate preparation completed!")
print("Total coordinate points:", len(coordinates_rad))
print("First 5 coordinates in radians:")
print(coordinates_rad[:5])
```

```python
earth_radius_km = 6371.0088
eps_km = 0.5
eps_radians = eps_km / earth_radius_km
dbscan = DBSCAN(
    eps=eps_radians,
    min_samples=10,
    metric="haversine",
    algorithm="ball_tree"
)
accidents["cluster"] = dbscan.fit_predict(coordinates_rad)
print("DBSCAN clustering completed!")
n_clusters = len(set(accidents["cluster"])) - (
    1 if -1 in accidents["cluster"].values else 0
)
n_noise = (accidents["cluster"] == -1).sum()
print("Number of clusters:", n_clusters)
print("Noise points:", n_noise)
print("Percentage of noise:", round(n_noise / len(accidents) * 100, 2), "%")
```

```python
from sklearn.cluster import DBSCAN
import pandas as pd
earth_radius_km = 6371.0088
eps_values = [0.5, 1, 2, 3, 5]
min_samples_values = [5, 10, 15]
results = []
for eps_km in eps_values:
    for min_samples in min_samples_values:
        model = DBSCAN(
            eps=eps_km / earth_radius_km,
            min_samples=min_samples,
            metric="haversine",
            algorithm="ball_tree"
        )
        labels = model.fit_predict(coordinates_rad)
        clusters = len(set(labels)) - (1 if -1 in labels else 0)
        noise = (labels == -1).sum()
        results.append({
            "eps_km": eps_km,
            "min_samples": min_samples,
            "clusters": clusters,
            "noise_points": noise,
            "noise_percentage": round(
                noise / len(labels) * 100, 2
            )
        })
results_df = pd.DataFrame(results)
display(
    results_df.sort_values(
        by="noise_percentage"
    ).reset_index(drop=True)
)
```

```python
best_model = DBSCAN(
    eps=2.0 / 6371.0088,
    min_samples=5,
    metric="haversine",
    algorithm="ball_tree"
)
accidents["cluster"] = best_model.fit_predict(coordinates_rad)
cluster_city_table = pd.crosstab(
    accidents["cluster"],
    accidents["city"]
)
print("CLUSTER VS CITY COMPARISON")
display(cluster_city_table)
print("\nNUMBER OF RECORDS IN EACH CLUSTER")
display(
    accidents["cluster"].value_counts().sort_index()
)
cluster_summary = accidents[
    accidents["cluster"] != -1
].groupby("cluster").agg(
    record_count=("accident_id", "count"),
    number_of_cities=("city", "nunique"),
    cities=("city", lambda x: ", ".join(sorted(x.unique()))),
    avg_latitude=("latitude", "mean"),
    avg_longitude=("longitude", "mean")
).reset_index()
print("\nCLUSTER SUMMARY")
display(cluster_summary)
```

```python
# STAGE 8: CITY-WISE DBSCAN EXPERIMENTS

from sklearn.cluster import DBSCAN
import numpy as np
import pandas as pd

earth_radius_km = 6371.0088

eps_values = [0.25, 0.5, 0.75, 1.0, 1.5]
min_samples_values = [5, 10]

city_results = []

for city in sorted(accidents["city"].unique()):

    city_data = accidents[
        accidents["city"] == city
    ].copy()

    city_coordinates = np.radians(
        city_data[["latitude", "longitude"]].to_numpy()
    )

    for eps_km in eps_values:
        for min_samples in min_samples_values:

            model = DBSCAN(
                eps=eps_km / earth_radius_km,
                min_samples=min_samples,
                metric="haversine",
                algorithm="ball_tree"
            )

            labels = model.fit_predict(city_coordinates)

            n_clusters = len(set(labels)) - (
                1 if -1 in labels else 0
            )

            n_noise = (labels == -1).sum()

            city_results.append({
                "city": city,
                "eps_km": eps_km,
                "min_samples": min_samples,
                "clusters": n_clusters,
                "noise_points": n_noise,
                "noise_percentage": round(
                    n_noise / len(labels) * 100, 2
                )
            })

city_results_df = pd.DataFrame(city_results)

display(
    city_results_df.sort_values(
        ["city", "eps_km", "min_samples"]
    ).reset_index(drop=True)
)
```

```python
comparison = city_results_df[
    (city_results_df["eps_km"].isin([0.75, 1.0, 1.5])) &
    (city_results_df["min_samples"] == 5)
].copy()

comparison = comparison.sort_values(
    ["city", "eps_km"]
).reset_index(drop=True)

display(comparison)
```

```python
# STAGE 10: K-DISTANCE PLOTS FOR EACH CITY

from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

earth_radius_km = 6371.0088
min_samples = 5

cities = sorted(accidents["city"].unique())

fig, axes = plt.subplots(4, 2, figsize=(15, 18))
axes = axes.flatten()

for i, city in enumerate(cities):

    city_data = accidents[
        accidents["city"] == city
    ]

    coords = np.radians(
        city_data[["latitude", "longitude"]].to_numpy()
    )

    neighbors = NearestNeighbors(
        n_neighbors=min_samples,
        metric="haversine",
        algorithm="ball_tree"
    )

    neighbors.fit(coords)

    distances, indices = neighbors.kneighbors(coords)
    k_distances = np.sort(
        distances[:, -1] * earth_radius_km
    )

    axes[i].plot(k_distances)
    axes[i].set_title(city)
    axes[i].set_xlabel("Points (sorted)")
    axes[i].set_ylabel("Distance to 5th neighbour (km)")
    axes[i].grid(True)

plt.suptitle(
    "K-Distance Plots for DBSCAN Parameter Selection",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
# STAGE 11: CLEAR K-DISTANCE PLOTS

from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

cities = sorted(accidents["city"].unique())

fig, axes = plt.subplots(4, 2, figsize=(16, 22))
axes = axes.flatten()

for i, city in enumerate(cities):

    city_data = accidents[accidents["city"] == city]

    coords = np.radians(
        city_data[["latitude", "longitude"]].to_numpy()
    )

    nn = NearestNeighbors(
        n_neighbors=5,
        metric="haversine",
        algorithm="ball_tree"
    )

    nn.fit(coords)
    distances, _ = nn.kneighbors(coords)

    k_distances = np.sort(
        distances[:, -1] * 6371.0088
    )

    axes[i].plot(k_distances)
    axes[i].set_title(city, fontsize=13)
    axes[i].set_xlabel("Sorted data points")
    axes[i].set_ylabel("5th-neighbour distance (km)")
    axes[i].grid(True)

plt.suptitle(
    "K-Distance Analysis for All 8 Cities",
    fontsize=17
)

plt.tight_layout()
plt.show()
```

```python
# STAGE 12: PUNE ACCIDENT CLUSTER MAP

import folium
import numpy as np
from sklearn.cluster import DBSCAN

# Select city
chosen_city = "Pune"

city_data = accidents[
    accidents["city"] == chosen_city
].copy()

# Prepare coordinates
coords = np.radians(
    city_data[["latitude", "longitude"]].to_numpy()
)

# Apply DBSCAN
model = DBSCAN(
    eps=1.0 / 6371.0088,
    min_samples=5,
    metric="haversine",
    algorithm="ball_tree"
)

city_data["cluster"] = model.fit_predict(coords)

# Create map centered on the city
center = [
    city_data["latitude"].mean(),
    city_data["longitude"].mean()
]

# Create map without external tile services

accident_map = folium.Map(
    location=center,
    zoom_start=11,
    tiles=None
)

# Add a simple background
folium.GeoJson(
    {
        "type": "FeatureCollection",
        "features": []
    }
).add_to(accident_map)

# Colors for clusters
colors = [
    "red", "blue", "green", "purple", "orange",
    "darkred", "darkblue", "cadetblue", "darkgreen",
    "pink", "black", "lightgreen"
]

# Plot accident records
for _, row in city_data.iterrows():

    cluster = row["cluster"]

    if cluster == -1:
        color = "gray"
    else:
        color = colors[cluster % len(colors)]

    folium.CircleMarker(
        location=[row["latitude"], row["longitude"]],
        radius=3,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.65,
        popup=f"Cluster: {cluster}"
    ).add_to(accident_map)
accident_map.save("Pune_Accident_Clusters.html")
display(accident_map)

accident_map.save("Pune_Accident_Clusters.html")
print("Pune cluster map created!")
print("Open Pune_Accident_Clusters.html from Colab Files.")
display(accident_map)
```

```python
# STAGE 13: PUNE CLUSTER ANALYSIS

# Recreate Pune dataset
city_data = accidents[
    accidents["city"] == "Pune"
].copy()

# Prepare coordinates
coords = np.radians(
    city_data[["latitude", "longitude"]].to_numpy()
)

# Run DBSCAN again
model = DBSCAN(
    eps=1.0 / 6371.0088,
    min_samples=5,
    metric="haversine",
    algorithm="ball_tree"
)

city_data["cluster"] = model.fit_predict(coords)

# Exclude noise points
pune_clusters = city_data[
    city_data["cluster"] != -1
].copy()

# Cluster statistics
cluster_summary = pune_clusters.groupby("cluster").agg(
    accident_records=("accident_id", "count"),
    avg_latitude=("latitude", "mean"),
    avg_longitude=("longitude", "mean"),
    avg_hour=("hour", "mean")
).reset_index()

print("PUNE CLUSTER SUMMARY")
display(cluster_summary)

# Severity analysis
severity_summary = pd.crosstab(
    pune_clusters["cluster"],
    pune_clusters["accident_severity"]
)

print("ACCIDENT SEVERITY BY CLUSTER")
display(severity_summary)

# Noise statistics
noise_count = (city_data["cluster"] == -1).sum()

print("Total Pune records:", len(city_data))
print("Noise points:", noise_count)
print("Clustered records:", len(pune_clusters))
```

```python
# STAGE 14: CLUSTER SIZE CHART

import matplotlib.pyplot as plt

cluster_summary.plot(
    x="cluster",
    y="accident_records",
    kind="bar",
    figsize=(10, 5),
    legend=False
)

plt.title("Pune - Accident Records per DBSCAN Cluster")
plt.xlabel("Cluster ID")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# STAGE 15: ANALYSIS OF ALL 8 CITIES

from sklearn.cluster import DBSCAN
import numpy as np
import pandas as pd

city_summary = []

for city in sorted(accidents["city"].unique()):

    city_data = accidents[
        accidents["city"] == city
    ].copy()

    coords = np.radians(
        city_data[["latitude", "longitude"]].to_numpy()
    )

    model = DBSCAN(
        eps=1.0 / 6371.0088,
        min_samples=5,
        metric="haversine",
        algorithm="ball_tree"
    )

    labels = model.fit_predict(coords)

    clusters = len(set(labels)) - (
        1 if -1 in labels else 0
    )

    noise = (labels == -1).sum()

    city_summary.append({
        "city": city,
        "total_records": len(city_data),
        "clusters": clusters,
        "noise_points": noise,
        "noise_percentage": round(
            noise / len(city_data) * 100, 2
        ),
        "clustered_records": len(city_data) - noise
    })

city_summary_df = pd.DataFrame(city_summary)

display(city_summary_df)
```

```python
# CITY-WISE CLUSTER COMPARISON

city_summary_df.plot(
    x="city",
    y="clusters",
    kind="bar",
    figsize=(10, 5),
    legend=False
)

plt.title("DBSCAN Cluster Count by City")
plt.xlabel("City")
plt.ylabel("Number of Clusters")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
city_summary_df.plot(
    x="city",
    y="noise_percentage",
    kind="bar",
    figsize=(10, 5),
    legend=False
)

plt.title("Noise Percentage by City")
plt.xlabel("City")
plt.ylabel("Noise Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
